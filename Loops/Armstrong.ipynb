```python
num = int(input("Enter the number: "))

a=num
NumberOfDigit=0
while(a!=0):
    a = a//10
    NumberOfDigit+=1

b=num
sum=0
while(b!=0):
    digit = b%10
    sum += digit**NumberOfDigit
    b = b//10

if(sum==num):
    print("The given number is Armstrong.")
else:
    print("The given number is not Armstrong.")
```
