```python
ch = input("Enter the character: ")
up=0
low=0
num=0
while(ch!='*'):
    if(ch.isupper()):
        up+=1
    elif(ch.islower()):
        low+=1
    elif(ch.isnumeric()):
        num+=1
    ch = input("Enter the character: ")

print("Uppercase: ", up)
print("Lowercase: ", low)
print("Numbers: ", num)
```
